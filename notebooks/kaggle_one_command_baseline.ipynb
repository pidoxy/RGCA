# RGCA Kaggle One-Command Baseline

This notebook avoids fragile multi-cell state. It delegates the full workflow to `scripts/kaggle_bootstrap_baseline.py`.

The script will:

1. find an existing `mimic_subset.jsonl`
2. build one from already-local PhysioNet files if possible
3. validate retrieval/eval split counts
4. run the structured experiment suite
5. summarize results into CSV/Markdown
6. package a private Kaggle dataset zip

No GCloud is used.


## 1. Setup Repo

Run this once. If Kaggle internet is off and the repo already exists, the pull may fail; in that case skip the pull and continue to Section 2.


````python
from pathlib import Path
import subprocess
import sys

PROJECT_ROOT = Path('/kaggle/working/RGCA')
print('PROJECT_ROOT exists:', PROJECT_ROOT.exists())

if not PROJECT_ROOT.exists():
    subprocess.run(['git', 'clone', 'https://github.com/pidoxy/RGCA.git', str(PROJECT_ROOT)], check=True)
else:
    subprocess.run(['git', '-C', str(PROJECT_ROOT), 'pull', 'origin', 'main'], check=True)

subprocess.run([sys.executable, '-m', 'pip', 'install', '-e', str(PROJECT_ROOT)], check=True)
````
